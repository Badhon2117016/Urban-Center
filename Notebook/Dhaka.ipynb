```python
# ============================================================
# CELL 6 — SENTINEL-2 COLLECTION CHECK FOR DHAKA
# Period: 2025-01-01 to 2025-03-31
# ============================================================

import ee

START_DATE = "2025-01-01"
END_DATE   = "2025-04-01"   # end date is exclusive
MAX_CLOUD  = 30             # scene-level cloud percentage

s2 = (
    ee.ImageCollection("COPERNICUS/S2_SR_HARMONIZED")
    .filterBounds(roi)
    .filterDate(START_DATE, END_DATE)
    .filter(ee.Filter.lte("CLOUDY_PIXEL_PERCENTAGE", MAX_CLOUD))
)

# Basic checks
scene_count = s2.size().getInfo()

dates = (
    s2.aggregate_array("system:time_start")
    .map(lambda t: ee.Date(t).format("YYYY-MM-dd"))
    .distinct()
    .getInfo()
)

print("✅ Sentinel-2 collection loaded")
print("Period:", START_DATE, "to 2025-03-31")
print("Cloud filter: <=", MAX_CLOUD, "%")

print("\nTotal Sentinel-2 scenes:", scene_count)
print("Unique acquisition dates:", len(dates))

print("\nFirst 15 dates:")
for d in dates[:15]:
    print(" -", d)

if scene_count == 0:
    raise ValueError("❌ No Sentinel-2 scenes found. Cloud threshold/date needs adjustment.")
else:
    print("\n✅ Sentinel-2 data available for Dhaka Division.")
```

```python
# ============================================================
# CELL 7 — CLOUD MASK + SENTINEL-2 MEDIAN COMPOSITE
# ============================================================

import ee

def mask_s2_clouds(img):
    """
    Cloud masking using SCL band.
    Removes:
    3  = cloud shadow
    8  = cloud medium probability
    9  = cloud high probability
    10 = cirrus
    11 = snow/ice
    """

    scl = img.select("SCL")

    mask = (
        scl.neq(3)
        .And(scl.neq(8))
        .And(scl.neq(9))
        .And(scl.neq(10))
        .And(scl.neq(11))
    )

    # Keep needed spectral bands
    spectral = img.select(
        ["B2", "B3", "B4", "B8", "B11", "B12"]
    )

    return (
        spectral
        .updateMask(mask)
        .divide(10000)
        .copyProperties(img, ["system:time_start"])
    )

# Apply cloud mask
s2_clean = s2.map(mask_s2_clouds)

# Create median composite
s2_composite = s2_clean.median().clip(roi)

print("✅ Cloud masking completed")
print("✅ Median composite created")

print("\nComposite bands:")
print(s2_composite.bandNames().getInfo())

print("\nExpected bands:")
print("B2  = Blue")
print("B3  = Green")
print("B4  = Red")
print("B8  = NIR")
print("B11 = SWIR1")
print("B12 = SWIR2")

print("\n✅ Sentinel-2 composite ready for index calculation.")
```

```python
# ============================================================
# CELL 8 — CALCULATE SENTINEL-2 INDICES
# NDBI, NDVI, MNDWI, BSI
# ============================================================

import ee

# ------------------------------------------------------------
# 1. NDVI = (NIR - Red) / (NIR + Red)
# ------------------------------------------------------------
ndvi = (
    s2_composite
    .normalizedDifference(["B8", "B4"])
    .rename("NDVI")
)

# ------------------------------------------------------------
# 2. NDBI = (SWIR1 - NIR) / (SWIR1 + NIR)
# ------------------------------------------------------------
ndbi = (
    s2_composite
    .normalizedDifference(["B11", "B8"])
    .rename("NDBI")
)

# ------------------------------------------------------------
# 3. MNDWI = (Green - SWIR1) / (Green + SWIR1)
# ------------------------------------------------------------
mndwi = (
    s2_composite
    .normalizedDifference(["B3", "B11"])
    .rename("MNDWI")
)

# ------------------------------------------------------------
# 4. BSI = ((SWIR1 + Red) - (NIR + Blue)) /
#          ((SWIR1 + Red) + (NIR + Blue))
# ------------------------------------------------------------
bsi = (
    s2_composite.expression(
        """
        ((SWIR + RED) - (NIR + BLUE)) /
        ((SWIR + RED) + (NIR + BLUE))
        """,
        {
            "SWIR": s2_composite.select("B11"),
            "RED":  s2_composite.select("B4"),
            "NIR":  s2_composite.select("B8"),
            "BLUE": s2_composite.select("B2")
        }
    )
    .rename("BSI")
)

# Combine all indices
s2_indices = (
    ee.Image.cat([
        ndvi,
        ndbi,
        mndwi,
        bsi
    ])
    .clip(roi)
)

print("✅ Sentinel-2 indices calculated successfully")

print("\nBands:")
print(s2_indices.bandNames().getInfo())

print("\nFormulas used:")
print("NDVI  = (B8 - B4) / (B8 + B4)")
print("NDBI  = (B11 - B8) / (B11 + B8)")
print("MNDWI = (B3 - B11) / (B3 + B11)")
print("BSI   = ((B11 + B4) - (B8 + B2)) / ((B11 + B4) + (B8 + B2))")

print("\n✅ Indices ready for built-up probability processing.")
```

```python
# ============================================================
# CELL 9 — BUILD RAW BUILT-UP SUITABILITY SCORE
# Using NDVI, NDBI, MNDWI and BSI
# ============================================================

import ee

# ------------------------------------------------------------
# Simple masks
# ------------------------------------------------------------

# Water-like areas
water_mask = mndwi.gt(0.0)

# Dense vegetation-like areas
vegetation_mask = ndvi.gt(0.45)

# ------------------------------------------------------------
# Normalize indices approximately from [-1, 1] to [0, 1]
# ------------------------------------------------------------

ndbi_n = ndbi.add(1).divide(2).clamp(0, 1)
bsi_n = bsi.add(1).divide(2).clamp(0, 1)

# Lower NDVI = more urban-like
ndvi_inverse = ee.Image(1).subtract(
    ndvi.add(1).divide(2).clamp(0, 1)
)

# Lower MNDWI = less water-like
mndwi_inverse = ee.Image(1).subtract(
    mndwi.add(1).divide(2).clamp(0, 1)
)

# ------------------------------------------------------------
# Raw built-up suitability
# ------------------------------------------------------------

builtup_raw = (
    ndbi_n.multiply(0.40)
    .add(bsi_n.multiply(0.25))
    .add(ndvi_inverse.multiply(0.20))
    .add(mndwi_inverse.multiply(0.15))
    .rename("Builtup_Raw")
)

# Remove obvious water and dense vegetation
builtup_raw_masked = (
    builtup_raw
    .updateMask(water_mask.Not())
    .updateMask(vegetation_mask.Not())
    .clip(roi)
    .rename("Builtup_Raw_Masked")
)

print("✅ Raw built-up suitability score created")
print("Band:", builtup_raw_masked.bandNames().getInfo())

print("\nWeights:")
print("NDBI        = 0.40")
print("BSI         = 0.25")
print("Inverse NDVI= 0.20")
print("Inverse MNDWI=0.15")

print("\nMasks:")
print("Water removed where MNDWI > 0.0")
print("Dense vegetation removed where NDVI > 0.45")

print("\n✅ Built-up suitability ready for visualization/check.")
```

```python
# ============================================================
# CELL 12 — VIIRS NIGHTTIME LIGHT FOR DHAKA
# 2025 Jan–Mar Monthly Composite
# ============================================================

import ee

# VIIRS monthly nighttime lights
viirs = (
    ee.ImageCollection("NOAA/VIIRS/DNB/MONTHLY_V1/VCMCFG")
    .filterBounds(roi)
    .filterDate("2025-01-01", "2025-04-01")
    .select("avg_rad")
)

# Check collection
viirs_count = viirs.size().getInfo()

print("VIIRS monthly images found:", viirs_count)

if viirs_count == 0:
    raise ValueError("❌ No VIIRS images found for this period.")

# Mean Jan–Mar nighttime light
viirs_mean = (
    viirs
    .mean()
    .clip(roi)
    .rename("VIIRS_Mean_Radiance")
)

# Basic statistics for Dhaka
stats = viirs_mean.reduceRegion(
    reducer=ee.Reducer.percentile([1, 5, 50, 95, 99])
        .combine(
            reducer2=ee.Reducer.minMax(),
            sharedInputs=True
        ),
    geometry=roi,
    scale=500,
    maxPixels=1e9,
    bestEffort=True
).getInfo()

print("\n✅ VIIRS Jan–Mar 2025 composite created")

print("\nDhaka VIIRS statistics:")
for key, value in stats.items():
    print(f"{key}: {value}")

print("\nBand:")
print(viirs_mean.bandNames().getInfo())

print("\n✅ VIIRS nighttime-light layer ready.")
```

```python
# ============================================================
# CELL 13 — CREATE VIIRS NIGHT LIGHT SCORE (0–1)
# Robust P5–P99 Normalization
# ============================================================

import ee

# Values obtained from your Dhaka VIIRS statistics
VIIRS_P5  = 0.6560206645786993
VIIRS_P99 = 19.527976172310964

# ------------------------------------------------------------
# Robust normalization
# P5  -> 0
# P99 -> 1
# Values outside range are clipped
# ------------------------------------------------------------

night_light_score = (
    viirs_mean
    .subtract(VIIRS_P5)
    .divide(VIIRS_P99 - VIIRS_P5)
    .clamp(0, 1)
    .rename("Night_Light_Score")
    .clip(roi)
)

# ------------------------------------------------------------
# Check resulting statistics
# ------------------------------------------------------------

score_stats = night_light_score.reduceRegion(
    reducer=ee.Reducer.percentile([5, 25, 50, 75, 90, 95, 99])
        .combine(
            reducer2=ee.Reducer.minMax(),
            sharedInputs=True
        ),
    geometry=roi,
    scale=500,
    maxPixels=1e9,
    bestEffort=True
).getInfo()

print("✅ VIIRS Night Light Score created")

print("\nNormalization:")
print("P5  =", VIIRS_P5, "→ 0")
print("P99 =", VIIRS_P99, "→ 1")
print("Values below P5  → 0")
print("Values above P99 → 1")

print("\nNight Light Score statistics:")

for key, value in score_stats.items():
    print(f"{key}: {value}")

print("\nBand:")
print(night_light_score.bandNames().getInfo())

print("\n✅ Night Light Score (0–1) ready.")
```

```python
# ============================================================
# CELL 15 — LOAD GHSL 2025 POPULATION CORRECTLY
# ============================================================

import ee

# Select the 2025 GHSL image directly by system:index
population = (
    ee.ImageCollection("JRC/GHSL/P2023A/GHS_POP")
    .filter(ee.Filter.eq("system:index", "2025"))
    .first()
    .select("population_count")
    .clip(roi)
    .rename("Population")
)

# Check population statistics over Dhaka
pop_stats = population.reduceRegion(
    reducer=ee.Reducer.percentile(
        [1, 5, 25, 50, 75, 90, 95, 99]
    ).combine(
        reducer2=ee.Reducer.minMax(),
        sharedInputs=True
    ),
    geometry=roi,
    scale=100,
    maxPixels=1e10,
    bestEffort=True
).getInfo()

print("✅ GHSL 2025 population loaded successfully")

print("\nPopulation statistics:")
for key, value in pop_stats.items():
    print(f"{key}: {value}")

print("\nBand:")
print(population.bandNames().getInfo())

print("\n✅ Population layer ready for scoring.")
```

```python
# ============================================================
# CELL 16 — CREATE POPULATION SCORE (0–1)
# Log transformation + P99 robust normalization
# ============================================================

import ee
import math

# From your actual Dhaka GHSL statistics
POP_P99 = 432.06758423846054

# ------------------------------------------------------------
# Log-normalized population score
#
# Population = 0   -> Score = 0
# Population = P99 -> Score = 1
# > P99            -> Score = 1
# ------------------------------------------------------------

population_score = (
    population
    .max(0)
    .add(1)
    .log()
    .divide(math.log(POP_P99 + 1))
    .clamp(0, 1)
    .rename("Population_Score")
    .clip(roi)
)

# ------------------------------------------------------------
# Check output statistics
# ------------------------------------------------------------

pop_score_stats = population_score.reduceRegion(
    reducer=ee.Reducer.percentile(
        [1, 5, 25, 50, 75, 90, 95, 99]
    ).combine(
        reducer2=ee.Reducer.minMax(),
        sharedInputs=True
    ),
    geometry=roi,
    scale=100,
    maxPixels=1e10,
    bestEffort=True
).getInfo()

print("✅ Population Score created successfully")

print("\nMethod:")
print("Log transformation + P99 normalization")
print("Population = 0       → Score 0")
print("Population ≈", round(POP_P99, 2), "→ Score 1")
print("Population > P99     → Score 1")

print("\nPopulation Score statistics:")

for key, value in pop_score_stats.items():
    print(f"{key}: {value}")

print("\nBand:")
print(population_score.bandNames().getInfo())

print("\n✅ Population Score (0–1) ready.")
```

```python
# ============================================================
# CELL 19 — CREATE WEIGHTED ROAD DENSITY
# FIXED VERSION
# 100 m analysis scale + 1 km neighborhood
# ============================================================

import ee

# ------------------------------------------------------------
# 1. Road-class weights
# ------------------------------------------------------------

ROAD_WEIGHTS = ee.Dictionary({
    "1": 1.00,
    "2": 0.85,
    "3": 0.70,
    "4": 0.45,
    "5": 0.25
})

def add_road_weight(feature):

    road_type = ee.Number(feature.get("GP_RTP")).format("%.0f")

    weight = ee.Number(
        ROAD_WEIGHTS.get(road_type, 0)
    )

    return feature.set("road_weight", weight)


weighted_roads = dhaka_roads.map(add_road_weight)


# ------------------------------------------------------------
# 2. Convert roads to weighted raster
# ------------------------------------------------------------

road_raster = (
    weighted_roads
    .reduceToImage(
        properties=["road_weight"],
        reducer=ee.Reducer.max()
    )
    .unmask(0)
    .rename("Road_Weight")
    .clip(roi)
)


# ------------------------------------------------------------
# 3. Calculate road density within 1 km
# ------------------------------------------------------------

road_density = (
    road_raster
    .reduceNeighborhood(
        reducer=ee.Reducer.mean(),
        kernel=ee.Kernel.circle(
            radius=1000,
            units="meters"
        )
    )
    .rename("Road_Density")
    .clip(roi)
)


# ------------------------------------------------------------
# 4. Statistics
# ------------------------------------------------------------

road_stats = road_density.reduceRegion(
    reducer=(
        ee.Reducer.percentile(
            [5, 25, 50, 75, 90, 95, 99]
        )
        .combine(
            reducer2=ee.Reducer.minMax(),
            sharedInputs=True
        )
    ),
    geometry=roi,
    scale=100,
    maxPixels=1e10,
    bestEffort=True,
    tileScale=4
).getInfo()


# ------------------------------------------------------------
# 5. Print results
# ------------------------------------------------------------

print("✅ Weighted road density created successfully")

print("\nRoad weights:")
print("Class 1 = 1.00")
print("Class 2 = 0.85")
print("Class 3 = 0.70")
print("Class 4 = 0.45")
print("Class 5 = 0.25")

print("\nNeighborhood radius = 1 km")
print("Analysis scale      = 100 m")

print("\nRoad Density statistics:")

for key, value in road_stats.items():
    print(f"{key}: {value}")

print("\nBand:")
print(road_density.bandNames().getInfo())

print("\n✅ Road density ready for normalization.")
```

```python
# ============================================================
# CELL 20 — CREATE ROAD DENSITY SCORE (0–1)
# Robust P5–P99 normalization
# ============================================================

import ee

# From your actual Dhaka road-density statistics
ROAD_P5  = 6.892915221714577e-05
ROAD_P99 = 0.21585093298869373

# ------------------------------------------------------------
# Robust normalization
# ------------------------------------------------------------

road_score = (
    road_density
    .subtract(ROAD_P5)
    .divide(ROAD_P99 - ROAD_P5)
    .clamp(0, 1)
    .rename("Road_Score")
    .clip(roi)
)

# ------------------------------------------------------------
# Check resulting statistics
# ------------------------------------------------------------

road_score_stats = road_score.reduceRegion(
    reducer=(
        ee.Reducer.percentile(
            [5, 25, 50, 75, 90, 95, 99]
        )
        .combine(
            reducer2=ee.Reducer.minMax(),
            sharedInputs=True
        )
    ),
    geometry=roi,
    scale=100,
    maxPixels=1e10,
    bestEffort=True,
    tileScale=4
).getInfo()

print("✅ Road Density Score created successfully")

print("\nNormalization:")
print("P5  =", ROAD_P5,  "→ Score 0")
print("P99 =", ROAD_P99, "→ Score 1")
print("Below P5  → 0")
print("Above P99 → 1")

print("\nRoad Score statistics:")

for key, value in road_score_stats.items():
    print(f"{key}: {value}")

print("\nBand:")
print(road_score.bandNames().getInfo())

print("\n✅ Road Score (0–1) ready.")
```

```python
# ============================================================
# CELL 24 — CREATE SMALL GRID CHUNKS OVER DHAKA DIVISION
# For chunk-wise Overture POI download
# ============================================================

import geopandas as gpd
from shapely.geometry import box
import math

# ------------------------------------------------------------
# Chunk size in degrees
# ~0.20 degree = roughly 20–22 km
# ------------------------------------------------------------

CHUNK_SIZE = 0.20

minx, miny, maxx, maxy = dhaka.total_bounds

print("Dhaka bounds:")
print("West :", minx)
print("South:", miny)
print("East :", maxx)
print("North:", maxy)

# ------------------------------------------------------------
# Build rectangular grid
# ------------------------------------------------------------

grid_cells = []

x = minx
chunk_id = 1

while x < maxx:
    y = miny

    while y < maxy:

        cell = box(
            x,
            y,
            min(x + CHUNK_SIZE, maxx),
            min(y + CHUNK_SIZE, maxy)
        )

        # Keep only cells that intersect Dhaka
        if cell.intersects(dhaka.geometry.iloc[0]):
            grid_cells.append({
                "chunk_id": chunk_id,
                "geometry": cell
            })
            chunk_id += 1

        y += CHUNK_SIZE

    x += CHUNK_SIZE

grid = gpd.GeoDataFrame(
    grid_cells,
    crs="EPSG:4326"
)

print("\n✅ Dhaka chunk grid created")
print("Chunk size:", CHUNK_SIZE, "degrees")
print("Total intersecting chunks:", len(grid))

# Show first 10 chunk bounds
print("\nFirst 10 chunks:")

for idx, row in grid.head(10).iterrows():

    b = row.geometry.bounds

    print(
        f"Chunk {row['chunk_id']}: "
        f"{b[0]:.4f}, {b[1]:.4f}, "
        f"{b[2]:.4f}, {b[3]:.4f}"
    )

print("\n✅ Grid ready for chunk-wise Overture download.")
```

```python
# ============================================================
# CELL 26 — DOWNLOAD + FILTER ALL DHAKA OVERTURE POI CHUNKS
# Storage-safe: each raw file is deleted immediately
# ============================================================

import os
import subprocess
import geopandas as gpd
import pandas as pd
import time

all_filtered_chunks = []

total_chunks = len(grid)

print(f"Starting full Dhaka POI processing for {total_chunks} chunks...\n")

for i, row in grid.iterrows():

    chunk_id = int(row["chunk_id"])
    minx, miny, maxx, maxy = row.geometry.bounds

    bbox_str = f"{minx},{miny},{maxx},{maxy}"

    raw_file = f"/content/overture_chunk_{chunk_id}_raw.geojson"

    # Remove old leftovers if any
    if os.path.exists(raw_file):
        os.remove(raw_file)

    state_file = raw_file + ".state"
    if os.path.exists(state_file):
        os.remove(state_file)

    print(f"[{chunk_id}/{total_chunks}] Downloading...")

    cmd = [
        "overturemaps",
        "download",
        "--bbox", bbox_str,
        "-f", "geojson",
        "--type", "place",
        "-o", raw_file
    ]

    result = subprocess.run(
        cmd,
        capture_output=True,
        text=True
    )

    # --------------------------------------------------------
    # If download failed, skip safely
    # --------------------------------------------------------
    if result.returncode != 0:

        print(f"   ❌ Chunk {chunk_id} failed")

        if os.path.exists(raw_file):
            os.remove(raw_file)

        if os.path.exists(state_file):
            os.remove(state_file)

        continue

    # --------------------------------------------------------
    # Read chunk
    # --------------------------------------------------------
    try:
        poi_chunk = gpd.read_file(raw_file)
    except Exception as e:
        print(f"   ❌ Read error: {e}")

        if os.path.exists(raw_file):
            os.remove(raw_file)

        if os.path.exists(state_file):
            os.remove(state_file)

        continue

    raw_count = len(poi_chunk)

    # --------------------------------------------------------
    # Urban category filter
    # --------------------------------------------------------
    if raw_count > 0:

        poi_chunk = poi_chunk[
            poi_chunk["basic_category"].isin(URBAN_POI_CATEGORIES)
        ].copy()

    filtered_count = len(poi_chunk)

    # --------------------------------------------------------
    # Clip to true Dhaka Division boundary
    # --------------------------------------------------------
    if filtered_count > 0:

        poi_chunk = gpd.sjoin(
            poi_chunk,
            dhaka[["geometry"]],
            predicate="within",
            how="inner"
        ).drop(
            columns=["index_right"],
            errors="ignore"
        )

    inside_count = len(poi_chunk)

    # --------------------------------------------------------
    # Keep only lightweight fields
    # --------------------------------------------------------
    if inside_count > 0:

        poi_chunk = poi_chunk[
            ["id", "basic_category", "confidence", "geometry"]
        ].copy()

        all_filtered_chunks.append(poi_chunk)

    # --------------------------------------------------------
    # Delete raw file immediately
    # --------------------------------------------------------
    if os.path.exists(raw_file):
        os.remove(raw_file)

    if os.path.exists(state_file):
        os.remove(state_file)

    print(
        f"   Raw: {raw_count} | "
        f"Urban: {filtered_count} | "
        f"Inside Dhaka: {inside_count}"
    )

    # Small pause
    time.sleep(0.2)


# ============================================================
# MERGE ALL FILTERED CHUNKS
# ============================================================

if len(all_filtered_chunks) == 0:
    raise RuntimeError("No POIs collected.")

dhaka_pois = gpd.GeoDataFrame(
    pd.concat(
        all_filtered_chunks,
        ignore_index=True
    ),
    crs="EPSG:4326"
)

# Remove duplicate Overture IDs
before_dedup = len(dhaka_pois)

dhaka_pois = (
    dhaka_pois
    .drop_duplicates(subset="id")
    .reset_index(drop=True)
)

after_dedup = len(dhaka_pois)

print("\n========================================")
print("✅ FULL DHAKA POI PROCESSING COMPLETE")
print("========================================")

print("Chunks processed       :", total_chunks)
print("POIs before duplicate removal:", before_dedup)
print("Unique urban POIs      :", after_dedup)
print("Duplicates removed     :", before_dedup - after_dedup)

print("\nTop 20 categories:")
print(
    dhaka_pois["basic_category"]
    .value_counts()
    .head(20)
)

print("\nGeometry types:")
print(dhaka_pois.geom_type.value_counts())

print("\n✅ Dhaka urban POI dataset ready.")
```

```python
# ============================================================
# CELL 33 — CREATE 100 m POI DENSITY RASTER
# CRS: EPSG:6933
# Neighborhood: 1 km
# ============================================================

import numpy as np
import geopandas as gpd
import rasterio
from rasterio.features import rasterize
from rasterio.transform import from_origin
from scipy.ndimage import uniform_filter

# ------------------------------------------------------------
# SETTINGS
# ------------------------------------------------------------

TARGET_CRS = "EPSG:6933"
CELL_SIZE = 100          # meters
NEIGHBORHOOD_M = 1000   # 1 km

print("========================================")
print("POI DENSITY PROCESSING")
print("========================================")

print("Input POIs       :", len(dhaka_pois))
print("Target CRS       :", TARGET_CRS)
print("Cell size        :", CELL_SIZE, "m")
print("Neighborhood     :", NEIGHBORHOOD_M, "m")


# ------------------------------------------------------------
# 1. Reproject Dhaka boundary + POIs
# ------------------------------------------------------------

dhaka_6933 = dhaka.to_crs(TARGET_CRS)
pois_6933 = dhaka_pois.to_crs(TARGET_CRS)

# Remove invalid / empty geometries
pois_6933 = pois_6933[
    pois_6933.geometry.notna()
    & ~pois_6933.geometry.is_empty
].copy()

print("\nValid POIs:", len(pois_6933))


# ------------------------------------------------------------
# 2. Build 100 m raster grid
# ------------------------------------------------------------

minx, miny, maxx, maxy = dhaka_6933.total_bounds

width = int(np.ceil((maxx - minx) / CELL_SIZE))
height = int(np.ceil((maxy - miny) / CELL_SIZE))

transform = from_origin(
    minx,
    maxy,
    CELL_SIZE,
    CELL_SIZE
)

print("\nRaster width :", width)
print("Raster height:", height)
print("Total cells  :", width * height)


# ------------------------------------------------------------
# 3. Rasterize Dhaka boundary mask
# ------------------------------------------------------------

boundary_mask = rasterize(
    [(geom, 1) for geom in dhaka_6933.geometry],
    out_shape=(height, width),
    transform=transform,
    fill=0,
    dtype="uint8"
)


# ------------------------------------------------------------
# 4. Rasterize POI counts
# Each POI contributes 1
# ------------------------------------------------------------

poi_count = rasterize(
    [(geom, 1) for geom in pois_6933.geometry],
    out_shape=(height, width),
    transform=transform,
    fill=0,
    merge_alg=rasterio.enums.MergeAlg.add,
    dtype="float32"
)

# Outside Dhaka = NaN later
print("\nPOI raster created")
print("Total rasterized POI count:",
      int(np.nansum(poi_count)))


# ------------------------------------------------------------
# 5. Calculate 1 km neighborhood POI density
#
# 100 m cells:
# 1 km radius ~ 10 cells
# Here we use a 21 x 21 moving window (~2.1 km across)
# ------------------------------------------------------------

radius_cells = int(NEIGHBORHOOD_M / CELL_SIZE)
window_size = radius_cells * 2 + 1

# Sum POIs in moving window
poi_density = uniform_filter(
    poi_count,
    size=window_size,
    mode="constant",
    cval=0
) * (window_size ** 2)

# Keep only Dhaka
poi_density = np.where(
    boundary_mask == 1,
    poi_density,
    np.nan
)


# ------------------------------------------------------------
# 6. Calculate statistics
# ------------------------------------------------------------

valid = poi_density[
    np.isfinite(poi_density)
]

percentiles = np.percentile(
    valid,
    [5, 25, 50, 75, 90, 95, 99]
)

print("\n========================================")
print("POI DENSITY STATISTICS")
print("========================================")

print("Min :", float(np.nanmin(valid)))
print("P5  :", float(percentiles[0]))
print("P25 :", float(percentiles[1]))
print("P50 :", float(percentiles[2]))
print("P75 :", float(percentiles[3]))
print("P90 :", float(percentiles[4]))
print("P95 :", float(percentiles[5]))
print("P99 :", float(percentiles[6]))
print("Max :", float(np.nanmax(valid)))

print("\nWindow size:", window_size, "x", window_size, "cells")

print("\n✅ POI density raster created.")
print("✅ Ready for robust 0–1 normalization.")
```

```python
# ============================================================
# CELL 34 — CREATE POI DENSITY SCORE (0–1)
# Log transformation + P99 normalization
# ============================================================

import numpy as np

# From your actual POI-density statistics
POI_P99 = 297.00000915527346

# ------------------------------------------------------------
# 1. Log transform + normalize
#
# Density = 0   -> Score 0
# Density = P99 -> Score 1
# > P99         -> Score 1
# ------------------------------------------------------------

poi_score = (
    np.log1p(np.nan_to_num(poi_density, nan=0.0))
    / np.log1p(POI_P99)
)

poi_score = np.clip(
    poi_score,
    0,
    1
)

# Outside Dhaka = NaN
poi_score = np.where(
    boundary_mask == 1,
    poi_score,
    np.nan
)

# ------------------------------------------------------------
# 2. Statistics
# ------------------------------------------------------------

valid_score = poi_score[
    np.isfinite(poi_score)
]

p = np.percentile(
    valid_score,
    [5, 25, 50, 75, 90, 95, 99]
)

print("========================================")
print("POI SCORE STATISTICS")
print("========================================")

print("Min :", float(np.nanmin(valid_score)))
print("P5  :", float(p[0]))
print("P25 :", float(p[1]))
print("P50 :", float(p[2]))
print("P75 :", float(p[3]))
print("P90 :", float(p[4]))
print("P95 :", float(p[5]))
print("P99 :", float(p[6]))
print("Max :", float(np.nanmax(valid_score)))

print("\nMethod:")
print("Log transformation + P99 normalization")
print("POI density = 0      -> Score 0")
print("POI density ≈ 297    -> Score 1")
print("POI density > P99    -> Score 1")

print("\n✅ POI Density Score (0–1) ready.")
```

```python
# ============================================================
# CELL 35A — FIX COPERNICUS DEM SLOPE
# Reproject DEM to metric CRS before slope calculation
# ============================================================

import ee

# ------------------------------------------------------------
# 1. Copernicus GLO-30 DEM
# ------------------------------------------------------------

dem_raw = (
    ee.ImageCollection("COPERNICUS/DEM/GLO30")
    .filterBounds(roi)
    .select("DEM")
    .mosaic()
    .clip(roi)
)

# ------------------------------------------------------------
# 2. Reproject to a metric CRS at 30 m
# EPSG:32646 = UTM Zone 46N
# Suitable for most of Dhaka Division
# ------------------------------------------------------------

dem_metric = (
    dem_raw
    .resample("bilinear")
    .reproject(
        crs="EPSG:32646",
        scale=30
    )
    .rename("Elevation")
)

# ------------------------------------------------------------
# 3. Calculate slope
# ------------------------------------------------------------

slope = (
    ee.Terrain.slope(dem_metric)
    .rename("Slope")
    .clip(roi)
)

# Keep DEM variable consistent for later steps
dem = dem_metric

# ------------------------------------------------------------
# 4. Check projection
# ------------------------------------------------------------

print("========================================")
print("DEM PROJECTION")
print("========================================")

print(dem.projection().getInfo())

# ------------------------------------------------------------
# 5. Recalculate slope statistics
# ------------------------------------------------------------

slope_stats_fixed = slope.reduceRegion(
    reducer=(
        ee.Reducer.percentile(
            [5, 25, 50, 75, 90, 95, 99]
        )
        .combine(
            reducer2=ee.Reducer.minMax(),
            sharedInputs=True
        )
    ),
    geometry=roi,
    scale=30,
    crs="EPSG:32646",
    maxPixels=1e10,
    bestEffort=True,
    tileScale=4
).getInfo()

print("\n========================================")
print("FIXED SLOPE STATISTICS")
print("========================================")

for key, value in slope_stats_fixed.items():
    print(f"{key}: {value}")

print("\nBand:")
print(slope.bandNames().getInfo())

# Check if valid
valid_values = [
    v for v in slope_stats_fixed.values()
    if v is not None
]

if len(valid_values) == 0:
    print("\n❌ Slope still has no valid values.")
else:
    print("\n✅ Slope calculation fixed successfully.")
```

```python
# ============================================================
# CELL 35B — CORRECT COPERNICUS GLO-30 SLOPE
# Calculate slope on each native DEM tile BEFORE mosaicking
# ============================================================

import ee

# ------------------------------------------------------------
# 1. Get only Copernicus DEM tiles intersecting Dhaka
# ------------------------------------------------------------

dem_collection = (
    ee.ImageCollection("COPERNICUS/DEM/GLO30")
    .filterBounds(roi)
)

print("DEM tiles intersecting Dhaka:",
      dem_collection.size().getInfo())


# ------------------------------------------------------------
# 2. Function: calculate slope on EACH tile in native projection
# ------------------------------------------------------------

def slope_from_native_tile(img):

    dem_tile = img.select("DEM")

    slope_tile = ee.Terrain.slope(
        dem_tile
    ).rename("Slope")

    return slope_tile.copyProperties(
        img,
        ["system:index"]
    )


# ------------------------------------------------------------
# 3. Calculate native-tile slopes
# ------------------------------------------------------------

slope_collection = dem_collection.map(
    slope_from_native_tile
)

# Mosaic AFTER slope calculation
slope = (
    slope_collection
    .mosaic()
    .clip(roi)
    .rename("Slope")
)

# DEM itself for later topographic scoring
dem = (
    dem_collection
    .select("DEM")
    .mosaic()
    .clip(roi)
    .rename("Elevation")
)


# ------------------------------------------------------------
# 4. Inspect one native DEM tile projection
# ------------------------------------------------------------

first_dem = ee.Image(dem_collection.first())

print("\n========================================")
print("NATIVE DEM TILE PROJECTION")
print("========================================")

print(
    first_dem
    .select("DEM")
    .projection()
    .getInfo()
)


# ------------------------------------------------------------
# 5. Correct slope statistics
# ------------------------------------------------------------

slope_stats_native = slope.reduceRegion(
    reducer=(
        ee.Reducer.percentile(
            [5, 25, 50, 75, 90, 95, 99]
        )
        .combine(
            reducer2=ee.Reducer.minMax(),
            sharedInputs=True
        )
    ),
    geometry=roi,
    scale=30,
    maxPixels=1e10,
    bestEffort=True,
    tileScale=4
).getInfo()


print("\n========================================")
print("CORRECTED SLOPE STATISTICS")
print("========================================")

for key, value in slope_stats_native.items():
    print(f"{key}: {value}")


print("\nBand:")
print(slope.bandNames().getInfo())


# ------------------------------------------------------------
# 6. Quick validation
# ------------------------------------------------------------

slope_max = slope_stats_native.get("Slope_max")

if slope_max is None:
    print("\n❌ No valid slope values.")

elif slope_max < 0.1:
    print(
        "\n⚠️ Slope is still suspiciously low. "
        "Do NOT create Topography Score yet."
    )

else:
    print("\n✅ Slope values now look physically plausible.")
    print("✅ Topography layer ready for scoring.")
```

```python
# ============================================================
# CELL 36 — CREATE TOPOGRAPHY SCORE (0–1)
# Low elevation + low slope = higher urban suitability
# ============================================================

import ee

# ------------------------------------------------------------
# Robust limits from your actual Dhaka statistics
# ------------------------------------------------------------

ELEV_P5  = 1.7209361412140318
ELEV_P95 = 18.246828582753512

SLOPE_P5  = 0.10552602599169365
SLOPE_P95 = 7.122808454748189

# ------------------------------------------------------------
# 1. Elevation suitability
# Lower elevation = higher score
# ------------------------------------------------------------

elevation_score = (
    ee.Image(1)
    .subtract(
        dem
        .subtract(ELEV_P5)
        .divide(ELEV_P95 - ELEV_P5)
        .clamp(0, 1)
    )
    .rename("Elevation_Score")
)

# ------------------------------------------------------------
# 2. Slope suitability
# Lower slope = higher score
# ------------------------------------------------------------

slope_score = (
    ee.Image(1)
    .subtract(
        slope
        .subtract(SLOPE_P5)
        .divide(SLOPE_P95 - SLOPE_P5)
        .clamp(0, 1)
    )
    .rename("Slope_Score")
)

# ------------------------------------------------------------
# 3. Combined Topography Score
# Give more weight to slope
# ------------------------------------------------------------

topography_score = (
    elevation_score.multiply(0.40)
    .add(
        slope_score.multiply(0.60)
    )
    .rename("Topography_Score")
    .clip(roi)
)

# ------------------------------------------------------------
# 4. Statistics
# ------------------------------------------------------------

topo_stats = topography_score.reduceRegion(
    reducer=(
        ee.Reducer.percentile(
            [5, 25, 50, 75, 90, 95, 99]
        )
        .combine(
            reducer2=ee.Reducer.minMax(),
            sharedInputs=True
        )
    ),
    geometry=roi,
    scale=100,
    maxPixels=1e10,
    bestEffort=True,
    tileScale=4
).getInfo()

print("========================================")
print("TOPOGRAPHY SCORE STATISTICS")
print("========================================")

for key, value in topo_stats.items():
    print(f"{key}: {value}")

print("\nWeights:")
print("Elevation suitability = 0.40")
print("Slope suitability     = 0.60")

print("\nInterpretation:")
print("Higher score = flatter + lower terrain")
print("Lower score  = steeper + higher terrain")

print("\nBand:")
print(topography_score.bandNames().getInfo())

print("\n✅ Topography Score (0–1) ready.")
```

```python
# ============================================================
# CELL 47 — THRESHOLD SENSITIVITY CHECK
# Compare P90, P92.5 and P95
# Same 10 ha minimum patch rule
# ============================================================

import numpy as np
from scipy import ndimage

# Percentile thresholds from final urban score
valid_score = urban_score[np.isfinite(urban_score)]

threshold_percentiles = [90, 92.5, 95]

structure = np.ones((3, 3), dtype="uint8")

print("========================================")
print("URBAN THRESHOLD SENSITIVITY")
print("========================================")

results = []

for perc in threshold_percentiles:

    threshold = np.percentile(valid_score, perc)

    # ----------------------------------------
    # Binary urban mask
    # ----------------------------------------
    binary = np.zeros_like(
        urban_score,
        dtype="uint8"
    )

    binary[
        np.isfinite(urban_score) &
        (urban_score >= threshold)
    ] = 1

    binary[~valid_mask] = 0

    # ----------------------------------------
    # Connected components
    # ----------------------------------------
    labels, n_initial = ndimage.label(
        binary,
        structure=structure
    )

    sizes = np.bincount(labels.ravel())

    if len(sizes) > 0:
        sizes[0] = 0

    # Minimum 10 ha
    keep = np.where(
        sizes >= 10
    )[0]

    cleaned = np.isin(
        labels,
        keep
    ).astype("uint8")

    cleaned[~valid_mask] = 0

    # ----------------------------------------
    # Final patches
    # ----------------------------------------
    clean_labels, _ = ndimage.label(
        cleaned,
        structure=structure
    )

    clean_sizes = np.bincount(
        clean_labels.ravel()
    )

    if len(clean_sizes) > 0:
        clean_sizes[0] = 0

    final_sizes = clean_sizes[
        clean_sizes > 0
    ]

    urban_pixels = int(
        cleaned.sum()
    )

    urban_area_km2 = (
        urban_pixels * 0.01
    )

    largest_ha = (
        float(final_sizes.max())
        if len(final_sizes) > 0
        else 0
    )

    median_ha = (
        float(np.median(final_sizes))
        if len(final_sizes) > 0
        else 0
    )

    results.append({
        "Percentile": perc,
        "Threshold": threshold,
        "Urban_km2": urban_area_km2,
        "Patches": len(final_sizes),
        "Largest_ha": largest_ha,
        "Median_ha": median_ha
    })


# ----------------------------------------
# Print comparison
# ----------------------------------------

for r in results:

    print(f"\nP{r['Percentile']}")

    print(
        "Threshold      :",
        round(r["Threshold"], 6)
    )

    print(
        "Urban area     :",
        round(r["Urban_km2"], 2),
        "km²"
    )

    print(
        "Final patches  :",
        r["Patches"]
    )

    print(
        "Largest patch  :",
        round(r["Largest_ha"], 2),
        "ha"
    )

    print(
        "Median patch   :",
        round(r["Median_ha"], 2),
        "ha"
    )

print("\n✅ Threshold comparison complete.")
```

```python
# ============================================================
# CELL 48 — FINAL URBAN CENTER OUTPUTS
# Threshold: P92.5
# Outputs:
# 1. Final Urban Score TIFF
# 2. Final Urban Center Mask TIFF
# 3. Final Urban Centers Shapefile
# ============================================================

import os
import numpy as np
import geopandas as gpd
import rasterio
from rasterio.features import shapes
from shapely.geometry import shape
from scipy import ndimage

# ------------------------------------------------------------
# SETTINGS
# ------------------------------------------------------------

FINAL_THRESHOLD = 0.660545
MIN_PATCH_HA = 10

OUTPUT_DIR = "/content/drive/MyDrive/GEE_Dhaka_Urban/Final_Outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)

print("========================================")
print("FINAL URBAN CENTER EXTRACTION")
print("========================================")

print("Threshold       :", FINAL_THRESHOLD)
print("Threshold basis : P92.5")
print("Minimum patch   :", MIN_PATCH_HA, "ha")

# ------------------------------------------------------------
# 1. Create binary mask
# ------------------------------------------------------------

final_binary = np.zeros_like(
    urban_score,
    dtype="uint8"
)

final_binary[
    np.isfinite(urban_score) &
    (urban_score >= FINAL_THRESHOLD)
] = 1

final_binary[~valid_mask] = 0

# ------------------------------------------------------------
# 2. Remove patches smaller than 10 ha
# 100 m pixel = 1 ha
# ------------------------------------------------------------

structure = np.ones((3, 3), dtype="uint8")

labels, n_labels = ndimage.label(
    final_binary,
    structure=structure
)

sizes = np.bincount(labels.ravel())

if len(sizes) > 0:
    sizes[0] = 0

keep_labels = np.where(
    sizes >= MIN_PATCH_HA
)[0]

final_urban_mask = np.isin(
    labels,
    keep_labels
).astype("uint8")

final_urban_mask[~valid_mask] = 0

# ------------------------------------------------------------
# 3. Save FINAL URBAN SCORE TIFF
# ------------------------------------------------------------

score_file = os.path.join(
    OUTPUT_DIR,
    "Dhaka_Final_Urban_Score_100m.tif"
)

score_profile = profile.copy()

score_profile.update(
    count=1,
    dtype="float32",
    nodata=-9999,
    compress="lzw"
)

score_to_write = np.where(
    np.isfinite(urban_score),
    urban_score,
    -9999
).astype("float32")

with rasterio.open(
    score_file,
    "w",
    **score_profile
) as dst:

    dst.write(
        score_to_write,
        1
    )

    dst.set_band_description(
        1,
        "Final_Urban_Score"
    )

print("\n✅ Urban Score TIFF saved")

# ------------------------------------------------------------
# 4. Save FINAL URBAN MASK TIFF
# ------------------------------------------------------------

mask_file = os.path.join(
    OUTPUT_DIR,
    "Dhaka_Final_Urban_Center_Mask_P92_5.tif"
)

mask_profile = profile.copy()

mask_profile.update(
    count=1,
    dtype="uint8",
    nodata=255,
    compress="lzw"
)

mask_to_write = final_urban_mask.copy()

# Outside Dhaka = NoData
mask_to_write[~valid_mask] = 255

with rasterio.open(
    mask_file,
    "w",
    **mask_profile
) as dst:

    dst.write(
        mask_to_write,
        1
    )

    dst.set_band_description(
        1,
        "Urban_Center"
    )

print("✅ Urban Center Mask TIFF saved")

# ------------------------------------------------------------
# 5. Polygonize urban mask
# ------------------------------------------------------------

polygon_records = []

for geom, value in shapes(
    final_urban_mask,
    mask=(final_urban_mask == 1),
    transform=transform
):

    if int(value) == 1:

        polygon_records.append({
            "geometry": shape(geom)
        })

urban_gdf = gpd.GeoDataFrame(
    polygon_records,
    crs="EPSG:6933"
)

print("\nInitial polygons:",
      len(urban_gdf))

# ------------------------------------------------------------
# 6. Calculate area
# ------------------------------------------------------------

urban_gdf["area_ha"] = (
    urban_gdf.geometry.area / 10000
)

urban_gdf["area_km2"] = (
    urban_gdf.geometry.area / 1e6
)

# Final safety filter
urban_gdf = urban_gdf[
    urban_gdf["area_ha"] >= MIN_PATCH_HA
].copy()

# Largest first
urban_gdf = urban_gdf.sort_values(
    "area_ha",
    ascending=False
).reset_index(drop=True)

# Urban center ID
urban_gdf["center_id"] = (
    np.arange(1, len(urban_gdf) + 1)
)

# ------------------------------------------------------------
# 7. Save shapefile
# ------------------------------------------------------------

shp_file = os.path.join(
    OUTPUT_DIR,
    "Dhaka_Final_Urban_Centers_P92_5.shp"
)

urban_gdf[
    [
        "center_id",
        "area_ha",
        "area_km2",
        "geometry"
    ]
].to_file(
    shp_file,
    driver="ESRI Shapefile"
)

print("✅ Urban Center Shapefile saved")

# ------------------------------------------------------------
# 8. Final statistics
# ------------------------------------------------------------

urban_pixels = int(
    np.sum(final_urban_mask == 1)
)

urban_area_km2 = (
    urban_pixels * 0.01
)

print("\n========================================")
print("FINAL OUTPUT SUMMARY")
print("========================================")

print("Final threshold    :", FINAL_THRESHOLD)
print("Urban pixels       :", urban_pixels)
print("Urban area         :", round(urban_area_km2, 2), "km²")
print("Urban centers      :", len(urban_gdf))

if len(urban_gdf) > 0:

    print(
        "Largest center     :",
        round(
            urban_gdf["area_km2"].max(),
            2
        ),
        "km²"
    )

    print(
        "Median center area :",
        round(
            urban_gdf["area_km2"].median(),
            2
        ),
        "km²"
    )

print("\nFiles saved in:")
print(OUTPUT_DIR)

print("\n1.", os.path.basename(score_file))
print("2.", os.path.basename(mask_file))
print("3.", os.path.basename(shp_file))

print("\n✅ FINAL DHAKA URBAN CENTER OUTPUTS READY.")
```

```python
# ============================================================
# CELL 49 — FIX FINAL URBAN CENTER SHAPEFILE
# Preserve 8-neighbour connected patches
# Expected: ~55 urban centers
# ============================================================

import os
import numpy as np
import geopandas as gpd
from rasterio.features import shapes
from shapely.geometry import shape
from scipy import ndimage

# ------------------------------------------------------------
# 1. Label final urban mask using 8-neighbour connectivity
# ------------------------------------------------------------

structure = np.ones((3, 3), dtype="uint8")

component_labels, component_count = ndimage.label(
    final_urban_mask == 1,
    structure=structure
)

# Pixel counts by component
component_sizes = np.bincount(
    component_labels.ravel()
)

component_sizes[0] = 0

# Keep only >= 10 ha
valid_component_ids = np.where(
    component_sizes >= MIN_PATCH_HA
)[0]

print("Connected components >= 10 ha:",
      len(valid_component_ids))


# ------------------------------------------------------------
# 2. Remove unwanted component labels
# ------------------------------------------------------------

labels_clean = np.where(
    np.isin(component_labels, valid_component_ids),
    component_labels,
    0
).astype("int32")


# ------------------------------------------------------------
# 3. Polygonize LABEL raster, not binary raster
# ------------------------------------------------------------

records = []

for geom, value in shapes(
    labels_clean,
    mask=(labels_clean > 0),
    transform=transform
):

    component_id = int(value)

    if component_id > 0:
        records.append({
            "component_id": component_id,
            "geometry": shape(geom)
        })


parts_gdf = gpd.GeoDataFrame(
    records,
    crs="EPSG:6933"
)

print("Polygon parts before dissolve:",
      len(parts_gdf))


# ------------------------------------------------------------
# 4. Dissolve all pieces belonging to same connected component
# ------------------------------------------------------------

urban_centers_fixed = (
    parts_gdf
    .dissolve(by="component_id")
    .reset_index()
)


# ------------------------------------------------------------
# 5. Calculate area
# ------------------------------------------------------------

urban_centers_fixed["area_ha"] = (
    urban_centers_fixed.geometry.area / 10000
)

urban_centers_fixed["area_km2"] = (
    urban_centers_fixed.geometry.area / 1e6
)

# Safety filter
urban_centers_fixed = urban_centers_fixed[
    urban_centers_fixed["area_ha"] >= MIN_PATCH_HA
].copy()

# Largest first
urban_centers_fixed = (
    urban_centers_fixed
    .sort_values(
        "area_ha",
        ascending=False
    )
    .reset_index(drop=True)
)

urban_centers_fixed["center_id"] = (
    np.arange(
        1,
        len(urban_centers_fixed) + 1
    )
)


# ------------------------------------------------------------
# 6. Save corrected shapefile
# ------------------------------------------------------------

fixed_shp = os.path.join(
    OUTPUT_DIR,
    "Dhaka_Final_Urban_Centers_P92_5_FIXED.shp"
)

urban_centers_fixed[
    [
        "center_id",
        "area_ha",
        "area_km2",
        "geometry"
    ]
].to_file(
    fixed_shp,
    driver="ESRI Shapefile"
)


# ------------------------------------------------------------
# 7. Final verification
# ------------------------------------------------------------

print("\n========================================")
print("CORRECTED URBAN CENTER SUMMARY")
print("========================================")

print(
    "Urban centers      :",
    len(urban_centers_fixed)
)

print(
    "Total urban area   :",
    round(
        urban_centers_fixed["area_km2"].sum(),
        2
    ),
    "km²"
)

print(
    "Largest center     :",
    round(
        urban_centers_fixed["area_km2"].max(),
        2
    ),
    "km²"
)

print(
    "Median center area :",
    round(
        urban_centers_fixed["area_km2"].median(),
        2
    ),
    "km²"
)

print("\nSaved:")
print(fixed_shp)

print(
    "\n✅ Corrected shapefile now follows "
    "8-neighbour connected urban patches."
)
```

```python
# ============================================================
# CELL 51 — REVERSE-GEOCODE URBAN CENTER POINTS
# OSM Nominatim, slow + respectful request rate
# ============================================================

import requests
import time
import pandas as pd

BASE_URL = "https://nominatim.openstreetmap.org/reverse"

headers = {
    "User-Agent": "DhakaUrbanCenterResearch/1.0"
}

name_records = []

print("Starting reverse geocoding for 55 urban centers...\n")

for idx, row in urban_points_wgs.iterrows():

    center_id = int(row["center_id"])
    lat = float(row["latitude"])
    lon = float(row["longitude"])

    params = {
        "lat": lat,
        "lon": lon,
        "format": "jsonv2",
        "zoom": 10,
        "addressdetails": 1
    }

    try:
        r = requests.get(
            BASE_URL,
            params=params,
            headers=headers,
            timeout=30
        )

        if r.status_code == 200:

            data = r.json()
            address = data.get("address", {})

            place_name = (
                address.get("city")
                or address.get("town")
                or address.get("municipality")
                or address.get("suburb")
                or address.get("village")
                or address.get("hamlet")
                or address.get("county")
                or data.get("display_name")
            )

            district = (
                address.get("county")
                or address.get("state_district")
                or ""
            )

            name_records.append({
                "center_id": center_id,
                "urban_name_raw": place_name,
                "district_raw": district,
                "display_name": data.get("display_name", "")
            })

            print(
                f"{center_id:02d} | "
                f"{place_name} | "
                f"{district}"
            )

        else:
            print(
                f"{center_id:02d} | FAILED | "
                f"HTTP {r.status_code}"
            )

            name_records.append({
                "center_id": center_id,
                "urban_name_raw": None,
                "district_raw": None,
                "display_name": None
            })

    except Exception as e:

        print(
            f"{center_id:02d} | ERROR | "
            f"{str(e)[:100]}"
        )

        name_records.append({
            "center_id": center_id,
            "urban_name_raw": None,
            "district_raw": None,
            "display_name": None
        })

    # Respect public Nominatim usage policy
    time.sleep(1.1)


# ------------------------------------------------------------
# Results table
# ------------------------------------------------------------

name_df = pd.DataFrame(name_records)

print("\n========================================")
print("REVERSE GEOCODING COMPLETE")
print("========================================")

print("Total centers :", len(name_df))
print(
    "Names found   :",
    name_df["urban_name_raw"].notna().sum()
)

print("\nFirst 20 results:")
print(
    name_df[
        [
            "center_id",
            "urban_name_raw",
            "district_raw"
        ]
    ].head(20).to_string(index=False)
)

print("\n✅ Place-name lookup complete.")
```
